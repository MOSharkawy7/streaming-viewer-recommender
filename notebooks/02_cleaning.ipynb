```python
import pandas as pd

ratings = pd.read_csv("../data/raw/ml-latest-small/ratings.csv")
movies  = pd.read_csv("../data/raw/ml-latest-small/movies.csv")

# Missing values per column
print(ratings.isna().sum())
print(movies.isna().sum())

# Duplicate rows
print("Duplicate rating rows:", ratings.duplicated().sum())
print("Same user rated same movie twice:", ratings.duplicated(subset=["userId", "movieId"]).sum())

# Data types
print(ratings.dtypes)

# Allowed rating values
print(ratings["rating"].value_counts().sort_index())
```

```python
ratings["rated_at"] = pd.to_datetime(ratings["timestamp"], unit="s")
ratings["rating_year"] = ratings["rated_at"].dt.year

print(ratings[["timestamp", "rated_at", "rating_year"]].head())
print(ratings["rated_at"].min(), ratings["rated_at"].max())
```

```python
movies["year"] = movies["title"].str.extract(r"\((\d{4})\)\s*$").astype("float")
movies["clean_title"] = movies["title"].str.replace(r"\s*\(\d{4}\)\s*$", "", regex=True)

print(movies[["title", "clean_title", "year"]].head())
print("Movies with no year found:", movies["year"].isna().sum())
```

```python
movies["genre_list"] = movies["genres"].str.split("|")

genre_counts = movies.explode("genre_list")["genre_list"].value_counts()
print(genre_counts)
```
